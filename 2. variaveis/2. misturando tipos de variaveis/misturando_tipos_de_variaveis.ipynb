- vamos aprender a como misturar tipos diferentes de variaveis
- no python não podemos printar ou realizar qualquer ação apenas colocando (), ele quebra
- como podemos reverter isso?
- a resposta é: FORMATANDO A INFORMAÇÃO
- str() int() etc....

- segue o exemplo abaixo


```python
# vamos dizer que nós precisamos criar um script de template para automatizar um lucro de uma empresa

faturamento = 3000
custo = 700
lucro = faturamento - custo

#para reverter a situação formatamos tudo em um tipo só que no caso é o string colocando str()
print("O faturamento da empresa é de " + str(faturamento));
print("E o custo foi de " + str(custo));
print("Então o Lucro foi de "+ str(lucro));
```
