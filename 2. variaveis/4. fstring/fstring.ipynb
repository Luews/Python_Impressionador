- Antes do texto você coloca um f"texto {variavel}" e dentro das chaves você coloca a variável que você quer

```python
faturamento = 5000
custo = 3100
lucro = faturamento - custo

print(f"O faturamento da empresa foi de {faturamento}, o custo foi de {custo}, e o lucro total foi de {lucro}")
```

```python
faturamento = float(input("Insira o Faturamento aqui: "))
custo =  float(input("Insira o custo aqui: "))
lucro = faturamento - custo
print(f"O lucro dessa operação foi de {lucro}")
```
