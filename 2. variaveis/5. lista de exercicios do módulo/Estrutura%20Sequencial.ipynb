#### 8. Vamos criar um conversor de temperatura. Faça um Programa que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.

#### 10. Tendo como dados de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, usando a seguinte fórmula:

##### a. Para homens: $P = 72,7h - 58$
##### b. Para mulheres: $P = 62,1h - 44,7$

```python
# fazer uma condição H = homens M = mulheres

condicaoSexo = input("Por favor digite M para medir peso masculino e F para feminino: ")


if condicaoSexo == "M":
    sexoM = float(input("Digite sua altura em Metros aqui: "))
    pesoIdealMasculino = (72.7 * sexoM - 58)
    print(f"Seu peso ideal para altura de {sexoM} é {pesoIdealMasculino: .2f} kgs")

else:
    sexoF = float(input("Digite sua altura em Metros aqui: "))
    pesoIdealFeminino = (62.1 * sexoF - 44.7)
    print(f"Seu peso ideal para altura de {sexoF} é {pesoIdealFeminino: .2f} kgs")



```

#####  Calcule o salário bruto (horas * salario por hora)

```python
ganhoHora = int(input("Quanto você ganha por hora?: ")) #ex: 9

horasMes = int(input("Quantas horas você trabalha no mês?: ")) #ex: 180

calculoSalario = horasMes * ganhoHora
print(f"Você tem o sálario de R${calculoSalario} por mês")

```

##### Calcule o desconto do IR (11% do salário bruto)

```python
ganhoHora = int(input("Quanto você ganha por hora?: ")) #ex: 9

horasMes = int(input("Quantas horas você trabalha no mês?: ")) #ex: 180

calculoIR = 0.11 * (ganhoHora * horasMes) #11% de:

print(f"O seu desconto de IR é de R${calculoIR} ")
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
ganhoHora = int(input("Quanto você ganha por hora?: ")) #ex: 9

horasMes = int(input("Quantas horas você trabalha no mês?: ")) #ex: 180

calculoINSS = 0.08 * (ganhoHora * horasMes)

print(f"O desconto do INSS no seu salário é de: R${calculoINSS}")
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
ganhoHora = int(input("Quanto você ganha por hora?: ")) #ex: 9

horasMes = int(input("Quantas horas você trabalha no mês?: ")) #ex: 180

calculoSind = 0.05 * (ganhoHora * horasMes)

print(f"O desconto do Singráficos é de: R${calculoSind}")
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
ganhoHora = int(input("Quanto você ganha por hora?: ")) #ex: 12

horasMes = int(input("Quantas horas você trabalha no mês?: ")) #ex: 180

salarioMes = ganhoHora * horasMes

calculoDescontos = (salarioMes) - calculoSind - calculoINSS - calculoIR

print(f"O seu sálario bruto é R${salarioMes} e o salário líquido é de R${calculoDescontos}")
```
