#### 1. Tamanho de strings. Faça um programa que leia 2 strings e informe o conteúdo delas seguido do seu comprimento. Informe também se as duas strings possuem o mesmo comprimento e são iguais ou diferentes no conteúdo.
<pre>
Compara duas strings
String 1: Brasil Hexa 2006
String 2: Brasil! Hexa 2006!
Tamanho de "Brasil Hexa 2006": 16 caracteres
Tamanho de "Brasil! Hexa 2006!": 18 caracteres
As duas strings são de tamanhos diferentes.
As duas strings possuem conteúdo diferente.
</pre>

```python
string1 = "Brasil Hexa 2026"
tamanhoS1 = len(string1)
string2 = "Brasil! Hexa 2026!"
tamanhoS2 = len(string2)


#objetivo colocar um é e a frase ficar:
#Brasil é Hexa 2026
print(f"{string1[:6]} é{string1[6:]}")

print(f"String 1: {string1}")
print(f"String 2: {string2}")
print(f"Tamanho de {string1}: {tamanhoS1} caracteres")
print(f"Tamanho de {string2}: {tamanhoS2} caracteres")

if tamanhoS1 == tamanhoS2:
    print("As duas strings são do mesmo tamanho")
else:
    print("As duas strings são de tamanhos diferentes")

#posso lançar um verificador se tem somente texto || isalpha( ) 
# no metodo isalpha( ) não entra ! @ #, apenas str A-Z

if string1 == string1.isalpha() and string2 == string2.isalpha():
    print("As duas strings possuem conteúdos iguais")
else:
    print("As duas strings possuem contéudos diferentes")
```

#### 2. Valida e corrige número de telefone. Faça um programa que leia um número de telefone, e corrija o número no caso deste conter somente 7 dígitos, acrescentando o '3' na frente. O usuário pode informar o número com ou sem o traço separador.
<pre>
Valida e corrige número de telefone
Telefone: 461-0133
Telefone possui 7 dígitos. Vou acrescentar o digito três na frente.
Telefone corrigido sem formatação: 34610133
Telefone corrigido com formatação: 3461-0133
</pre>

```python
numeroT = input("Digite um número de telefone:")

#tratar o numero do usuario se colocar - 
#espaços
numeroT = numeroT.strip()
#trata o -
numeroT = numeroT.replace('-', '')

print(numeroT)

if len(numeroT) == 7:
    print('Telefone possui 7 dígitos. Vou acrescentar o digito três por último.')
    numeroT = numeroT + '3'
    #dessa forma que a gente pode usar o append no python
    #variavel = 'string' + variavel (kkkkk simples assim)
    print(f"Número corrigido com formatação {numeroT[:4]}-{numeroT[4:]}")
else:
    print(f"O Número informado em questão formatado é:{numeroT[:4]}-{numeroT[4:]}")



```
